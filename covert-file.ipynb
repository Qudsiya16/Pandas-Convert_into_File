#Convert the raw data into File(csv, excel, json)

```python
import pandas as pd
```

```python
#dictionary data first converted into tabular form then into any type of file.
Record = {
    "Roll no" : [221,222,223,224,225,226,227],
    "Name": ["Ali", "Sara", "Zainab", "Mahrosh", "Kinza", "Maham", "Rimsha"],
    "Maths" :[20, 30, 40, 50, 60, 70, 80],
    "Science":[25, 35, 45, 55, 65, 75, 85]
}
a=pd.DataFrame(Record)
a
```

```python
#Code to change tabular data into any file:
a.to_excel("Record.xlsx", index=False)
```

```python
#to read excel file
b = pd.read_excel("Record.xlsx")
b
```
